# Note 1: UniProt
- "UniProt" -> an open-access database of protein seq and func information

# Note 2: NCBI & UCSC
- Browsers/databases for retrieving gene sequences (Genome Browsers)
- What is "RefSeq"? -> 
- "NM" -> mRNA, can be seen in RefSeq transcripts
- "NP" -> Pr, can be seen in RefSeq proteins
- "NG" -> Gene, can be seen in RefSeqGene
- NCBI PubMed can be used to find articles related to a specific gene or protein.

# Note 3: Pairwise sequence alignment
- Sequence comparison through alignment enables the identification of conserved patterns and sequence variations.
- Sequence alignment is fundamental to predicting the structure and function of unknown sequences.
- 2 types: Global, Local
