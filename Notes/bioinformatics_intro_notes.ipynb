# Note 4: PAM & BLOSUM Matrices
- PAM: Point Accepted Mutation, BLOSUM: BLOcks SUbstitution Matrix
- Two types of **Substitution Matrices** for aminoacids,

    PAM and BLOSUM essentially capture evolutionary history by using observed amino-acid substitution patterns in homologous proteins to determine which substitutions are more common or rarer during evolution.

- Alignment therefore accounts for the fact that some substitutions are more evolutionarily acceptable than others, rather than simply treating all mismatches as equally unfavorable.
- The PAM matrix is based on observed evolutionary substitutions in homologous proteins.

    Unlike PAM, BLOSUM matrices are derived directly from conserved blocks of homologous proteins.
- PAM1 represents a low level of evolutionary divergence, classically corresponding to approximately 1 accepted mutation per 100 amino-acid positions.

    PAM250 is derived by raising the PAM1 mutation probability matrix to the 250th power.

    Larger PAM matrices are used for greater evolutionary distances, whereas smaller PAM matrices are used for closer evolutionary relationships.

- Unlike PAM, higher BLOSUM numbers are used for more closely related sequences.

    Lower BLOSUM numbers are used for more distantly related sequences.

# Note 6: PSSM
- Motif: A short, recurring sequence pattern in DNA, RNA, or proteins that is typically associated with a specific function or biological role
**A sequence pattern expected at transcription factor (TF) binding sites.**
- Background nucleotide frequency: The expected probability of observing each nucleotide in a genomic region in the absence of a transcription factor binding motif.

- PSSM: Position-Specific Scoring Matrices
- A table representing the probability or log-odds score of each amino acid or nucleotide at each position in a gap-free multiple sequence alignment.

    Rows → Residue identities

    Columns → Residue positions in the multiple sequence alignment

    Values → Log-odds scores for each residue at each position

- A PSSM asks: “How significant is observing this nucleotide at this position compared with its expected frequency in the background?”

That's why we have:
$$
PSSM(b,i) = \log_2\left(\frac{P(b \mid motif,i)}{P(b \mid background)}\right)
$$

- Making PSSM:

    Count Matrix -> Probability: PWM -> log(2): PSSM
- first PWM (Position-Weight Matrix) matrix in Python => PSSM matrix

# Note 7: Phylogenetics
- **Phylogenetic analysis:** Study the evolutionary history of organisms using phylogenetic trees.
- **Sequence homology:** Molecular sequences used to construct phylogenetic trees are assumed to be homologous, i.e., derived from a common ancestor.
- **Taxon:** Any species or sample represented at the tip (terminal node) of a phylogenetic tree. The plural form is **taxa**.
- **Clade:** A group of taxa derived from a common ancestor; it consists of a node and all of its descendants.
- **Phylogenetic tree-building methods:** UPGMA, Neighbor-Joining (NJ), Maximum Parsimony, Maximum Likelihood, etc.
- **Phylogenetic tree software:** MEGA (Molecular Evolutionary Genetics Analysis).
-
```text
         ┌── A
     ┌───┤
     │   └── B
  ───┤              -> This tree can also be stored in Newick format -> ((A,B),(C,D));
     │   ┌── C         or ((A:0.1,B:0.2):0.3,(C:0.4,D:0.1):0.2);
     └───┤             the numbers are each branch length
         └── D
```

- formats:

    PhyloXML -> richer metadata

## Phylogenetic Methods

Phylogenetic tree-building methods can be divided into three main categories:

### 1. Distance-based Methods

First calculate a **Distance Matrix**, then construct the tree.

* **UPGMA** — assumes a molecular clock.
* **Neighbor-Joining (NJ)** — does not require a molecular clock.

```text
MSA → Distance Matrix → Tree
```

### 2. Character-based Methods

Work directly with **character patterns in the alignment**.

* **Maximum Parsimony (MP)** — searches for the tree requiring the fewest evolutionary changes.

### 3. Model-based Methods

Use an explicit **evolutionary model** to estimate how sequences have changed.

* **Maximum Likelihood (ML)**
* **Bayesian methods**

> **Key distinction:** Distance-based methods reduce the alignment to pairwise distances, whereas character-based and model-based methods use the information in the alignment more directly.

Biopython is useful for learning and programmatic phylogenetic workflows, but professional phylogenetic analysis often uses specialized tools such as **IQ-TREE, RAxML, and MrBayes**.
