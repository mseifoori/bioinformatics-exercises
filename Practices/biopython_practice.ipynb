```python
from Bio import Align

seq_1 = 'AAGCGA'
seq_2 = 'AAC'

aligner = Align.PairwiseAligner()

aligner.mode = 'local'

alignments = aligner.align(seq_1, seq_2)

len(alignments)
```

```python
for alignment in alignments:
    print(f'{alignment}\n\nAlignment Score:{aligner.score}')
```
